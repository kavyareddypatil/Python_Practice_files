```python
#GetDummies
data = {"firstname" : ["Kavya","Lahari","Muni","Anitha","Mahitha"],
        "lastname" : ["Patil","Tavasi","Sharma","Devi","Ammuri"],
        "Department" : ["MCA","EEE","MBA","CSE","ECE"],
        "Country" : ["India","USA","UK","Germany","Australia"]}
details = pd.DataFrame(data,columns=["firstname","lastname","Department","Country"])
print(details)
```

```python
df1 = pd.get_dummies(details["Department"],dtype = "int")
df1
```

```python
df2 = pd.get_dummies(details["Country"],dtype = "int")
df2
```

```python
#remove duplicates from the data by using drop_duplicates
import pandas as pd
info = pd.DataFrame({"Name":["John","Benny","Allice","Smith","Benny"],
                     "Age":[25,26,35,29,26]})
print(info)
print("*"*25)
info = info.drop_duplicates()
print(info)
```

```python
employee = pd.DataFrame({"ID":[101,102,105,107,102,103,101],
                         "Department":["IT","HR","Finance","HR","HR","Finance","IT"],
                         "Salary":[40000,50000,45000,60000,50000,57000,40000]
                         })
print(employee)
print("-"*30)
remove_duplicates = employee.drop_duplicates()
print(remove_duplicates)
```

```python
# apply() function in side function
def total_sal(salary):
  increment = 1000
  return salary+increment
data = pd.DataFrame({
    "dept1_salary" : [25000,30000,45000],
    "dept2_salary" : [30000,27000,50000]
})
print(data)
print("="*30)
print("After Salary increment")
df = data.apply(total_sal)
print(df)


```

```python
def total_marks(marks):
  assignment_marks = 10
  return marks + assignment_marks
data = pd.DataFrame({"section_A":[25,30,27,29,26],
                     "section_B":[30,21,26,29,27]})
print(data)
print("*"*25)
print("After assignment marks added to the marks")
df = data.apply(total_marks)
print(df)
```

```python
import pandas as pd
data= pd.DataFrame({"Name":["John","Alice","Bob","Jenny"],
                    "Gender":["Male","Female","Male","Female"],
                    "Group":["CSE","MCA","CSE","ECE"],
                    "Age":[21,22,20,25]})
print(data)
print("-"*30)
df1 = pd.Categorical(data["Group"])
print(df1)
df2 = pd.Categorical(data["Gender"])
print(df2)
```
